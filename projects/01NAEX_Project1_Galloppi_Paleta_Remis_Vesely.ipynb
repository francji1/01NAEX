## Background

For small release angles the period of a simple pendulum of length $L$ is
$$
T_0 = 2\pi\sqrt{\frac{L}{g}},
$$
independent of the mass of the bob. For a release angle $\theta_0$ (in radians) the period is longer,
$$
T(\theta_0) \approx T_0\left(1+\frac{\theta_0^2}{16}+\frac{11\,\theta_0^4}{3072}\right),
\qquad \log T = \log(2\pi) - \tfrac12\log g + \tfrac12 \log L + x(\theta_0),
\qquad x(\theta_0) = \log\left(1+\frac{\theta_0^2}{16}+\frac{11\,\theta_0^4}{3072}\right).
$$

| $\theta_0$ | 10° | 20° | 30° | 45° | 50° | 60° |
|---|---|---|---|---|---|---|
| $T/T_0 - 1$ | 0.19 % | 0.77 % | 1.74 % | 4.0 % | 5.0 % | 7.3 % |

For $L = 0.65$ m ($T_0 = 1.62$ s) the period at 30° is longer by only 0.025 s than at 10°, at 50° by 0.077 s.

Note the structure: on the $T$ scale the effects of $L$ and $\theta_0$ **multiply** (the angle adds about 5 % to $T$, which is
0.06 s for $L = 0.35$ m but 0.10 s for $L = 1$ m, so there is an $L \times \theta_0$ interaction), while on the $\log T$
scale they **add** and there is no interaction. Keep this in mind when you choose the scale for a design that assumes no
interactions.

Things that make a real pendulum differ from the ideal one:

* $L$ is the distance from the pivot to the **centre of mass** of the bob. For a ring threaded on the string this is the length of
  the string from the pivot to the point where the ring hangs plus the (inner) radius of the ring. Several identical rings
  threaded at the same point change the mass but not the centre of mass.
* A ring is not a point mass. A thin ring of radius $R$ swinging in its own plane adds about $R^2/L$ to the effective length
  ($R^2/(2L)$ if it swings perpendicular to its plane); this is the same for 1, 2 or 3 identical rings.
* The amplitude decreases during the timed oscillations (air drag), more for a light bob.
* The pivot must not move; the string must not stretch or twist; the string should be much lighter than one ring.
* A hand-operated stopwatch has a reaction-time error of roughly 0.05-0.2 s at the start and at the stop, and each operator has
  his or her own habits. Timing $N$ oscillations divides this error by $N$.

Last year most teams used angles of 5-20° and reported that "the angle has no effect" and that "mass is not significant,
therefore it does not matter". Neither conclusion followed from their data: the angle effect was too small for their precision,
and a non-significant test is not evidence of no effect.

## The experiment

### Equipment and factors

* **Three pendulums** hung in advance from one rigid bar (or a door frame), at three lengths across your feasible range,
  e.g. about 0.35, 0.65 and 1.0 m. Measure each $L$ (pivot to the centre of the ring) carefully once and describe how.
  Changing the length then means moving the rings to another pendulum, which takes seconds and lets you randomize $L$ fully.
* **Mass:** 1, 2 or 3 identical rings (or nuts) threaded at the same point. **Weigh** each configuration and record the mass in
  grams: the mass is a factor with 3 levels and, at the same time, a numeric variable.
* **Release angle:** 3 levels, at least one $\le 10°$ and at least one $\ge 45°$, e.g. 10°, 30° and 50°. Set it by the horizontal
  displacement $d = L\sin\theta_0$ (a ruler on the floor or a table) or by a protractor; record the angle you actually used.
* **Response:** the period $T = t_N/N$ from the time $t_N$ of $N$ full oscillations, with the **same $N$ in all runs**
  ($N = 20$ is recommended). Release the bob, let it settle for one or two swings, start timing at a pass through the lowest
  point. Record the raw $t_N$, not only $T$.
* **Operator** = the person who times the run. Only **one** person times each run; the others release the bob, count and record.
  (If several people time the same swing, these are repeated measurements of one run, not replicates; we will meet this
  distinction in Lecture 11.)

### Part 0: pilot study (10 runs)

One operator times 10 runs at one setting: the middle length, 30°, 2 rings. From these you estimate the run-to-run standard
deviation $\sigma$ of $T$ and choose the number of blocks in Part A.

### Part A: factorial in randomized complete blocks (RCBD, Lecture 3)

* Treatments: the $3 \times 3 = 9$ combinations of length and angle; the mass is fixed (2 rings).
* Block = operator: one operator times all 9 runs of a block in a random order (random order within each block).
* Number of blocks $b$ from the pilot study (Task 2), **at least 3**. In a team of two, one operator times two blocks
  (a block is then an operator in a session); in a team of four, each member can time one block.
* This design can estimate the $L \times \theta_0$ interaction.

### Part B: two Graeco-Latin squares (Lecture 4)

* A $3 \times 3$ Graeco-Latin square with **rows = operator, columns = mass** (1, 2, 3 rings), **Latin letters = angle**
  and **Greek letters = length**: 9 runs, each operator times every mass once, every angle once and every length once.
* In a team of two, the rows are three **rounds** (three consecutive groups of three runs) timed by one operator, and the other
  operator times the second square.
* Run **two squares**, each randomized separately (random permutation of rows, columns and both sets of symbols, random run
  order of the 9 runs within the square). One square leaves no degrees of freedom for error in the factor model (why?).
* This design estimates four main effects from 9 runs, but it assumes that there are **no interactions**.

### Recording

For every run record: `part` (pilot / A / B), `block` (Part A, otherwise 0), `square` and `row` (Part B, otherwise 0),
`run` (the order in which the runs were actually done, over the whole session), `operator`, `date_time`, `L_m`, `n_rings`,
`mass_g`, `theta_deg`, `N`, `t_N_s` and a `note` on anything unusual (missed count, the bob hit something, ...). Do not delete
runs; mark them and explain.

## Helper functions

You may use these functions (or write your own).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.graphics.factorplots import interaction_plot
from pathlib import Path
from statsmodels.stats.stattools import durbin_watson


def anova_power(lam, df1, df2, alpha=0.05):
    """Power of the ANOVA F-test: P(F'(df1, df2, lam) > F_{1-alpha; df1, df2})."""
    return stats.ncf.sf(stats.f.ppf(1 - alpha, df1, df2), df1, df2, lam)


def tukey_from_model(model, data, factor, response, alpha=0.05):
    """Tukey's HSD for the levels of `factor`, using MS_E and df_E of a fitted model (balanced data)."""
    mse, dfe = model.mse_resid, model.df_resid
    means = data.groupby(factor, observed=True)[response].mean()
    n = data.groupby(factor, observed=True)[response].count().iloc[0]
    k = len(means)
    se = np.sqrt(mse / n)
    hsd = stats.studentized_range.ppf(1 - alpha, k, dfe) * se
    rows = []
    levels = list(means.index)
    for i in range(k):
        for j in range(i + 1, k):
            diff = means.iloc[j] - means.iloc[i]
            rows.append({"pair": f"{levels[j]} - {levels[i]}", "diff": diff,
                         "lower": diff - hsd, "upper": diff + hsd,
                         "p_adj": stats.studentized_range.sf(abs(diff) / se, k, dfe),
                         "reject": abs(diff) > hsd})
    print(f"MS_E = {mse:.4g}, df_E = {dfe:.0f}, n per level = {n}, HSD = {hsd:.4g}")
    return pd.DataFrame(rows)


def theory_period(L, theta_deg, g=9.81):
    """Period of a simple pendulum of length L (m) released at theta_deg degrees (series to the 4th order)."""
    th = np.radians(theta_deg)
    return 2 * np.pi * np.sqrt(L / g) * (1 + th**2 / 16 + 11 * th**4 / 3072)


def x_angle(theta_deg):
    """Large-angle correction on the log scale: x = log(1 + th^2/16 + 11 th^4/3072), th in radians."""
    th = np.radians(theta_deg)
    return np.log(1 + th**2 / 16 + 11 * th**4 / 3072)


def random_graeco_latin_3(seed=None):
    """Randomized 3 x 3 Graeco-Latin square in long format (row, col, latin, greek; levels coded 1, 2, 3).

    Two orthogonal cyclic squares with randomly permuted rows, columns, Latin and Greek symbols;
    the column `order` is a random run order of the 9 cells.
    """
    rng = np.random.default_rng(seed)
    i, j = np.meshgrid(np.arange(3), np.arange(3), indexing="ij")
    latin, greek = (i + j) % 3, (i + 2 * j) % 3
    r, c = rng.permutation(3), rng.permutation(3)
    latin, greek = rng.permutation(3)[latin[r][:, c]], rng.permutation(3)[greek[r][:, c]]
    sheet = pd.DataFrame({"row": i.ravel() + 1, "col": j.ravel() + 1,
                          "latin": latin.ravel() + 1, "greek": greek.ravel() + 1})
    sheet["order"] = rng.permutation(9) + 1
    return sheet
```

**Part A** model: $T \sim \text{block} + L \times \theta_0$
- every angle level is observed $3\cdot b$ times $\ldots$ 3 lengths x b blocks

```python
D = theory_period(0.65, 10, g=9.81) - theory_period(0.65, 30, g=9.81)
df1 = 2
power_results = []

for b in [2, 3, 4, 5, 6]:
    df2 = 8 * (b - 1)
    lam = 3*b * D**2 / (2 * (pilot_std**2))
    power = anova_power(lam, df1, df2, alpha=0.05)
    power_results.append({
        "Blocks (b)": b,
        "Power": f"{power:.2%}"
    })

power_df = pd.DataFrame(power_results)
display(power_df.set_index('Blocks (b)'))
```

**5a: ANOVA and the choice of scale of response variable T**

Summary - model: $y_{ijk} = \mu + \beta_k + \lambda_i + \tau_j + (\lambda\tau)_{ij} + \varepsilon_{ijk}$,

where $\beta_k$ is the block (operator), $\lambda_i$ the length, $\tau_j$ the angle; 9 length × angle combinations × 3 operators = 27 runs, $df_E = 26 - 2 - 2 - 2 - 4 = 16$.

The `length` and the `angle` are highly significant on both scales: we reject $H_0$ (all level effects of the factor are zero). The interaction is not significant at 5 % on either scale, but its $F$ statistic is ten times larger on the $T$ scale ($p \approx 0.09$) than on the $\log T$ scale ($p \approx 0.9$).

**Why the interaction appears for $T$ and disappears for $\log T$.**

Physically $T = 2\pi\sqrt{L/g}\,\bigl(1 + \theta_0^2/16 + \dots\bigr)$ (Taylor series): the angle multiplies the small-angle period $T_0 = 2\pi\sqrt{L/g}$ by a factor $f(\theta_0)$ that does not depend on $L$.
On the $T$ scale the angle therefore adds the same *percentage*, i.e. more seconds for a long pendulum; the effect of the angle depends on $L$, which is an interaction.

The (natural) logarithm turns the product into a sum, $\ln T = \ln 2\pi - \tfrac12\ln g + \tfrac12\ln L + x(\theta_0)$, where $x(\theta_0) = \ln f(\theta_0) \approx \theta_0^2/16$,
so the angle adds the same amount $x(\theta_0)$ at every length $L$ and the interaction vanishes.

**Why did the test not detect the interaction?**

With $df_E = 16$ and $\hat\sigma \approx 0.013$ s, our experiment has little power for the interaction on the $T$ scale. Its theoretical size, i.e. the difference between the angle effect (5° → 45°) at 1.00 m and at 0.35 m, is only about $0.03$ s (using only the term $\theta_0^2/16$). So $p \approx 0.09$ does not mean that the interaction is absent; on the $\ln T$ scale it is absent in theory.

**Preferred scale: $\ln T$.**

The model is additive (no interaction needed) and the effects have a physical meaning: the angle effect $x(\theta_0)$ reads directly as a relative change of $T$, and when $\ln L$ is used as a continuous regressor its coefficient is the exponent 1/2.

```python
# --- Task 5b: Tukey's HSD with MS_E and df_E of our model, comparison with the theory ---
A["T_theory"] = theory_period(A["L_m"], A["theta_deg"])
A["logT_theory"] = np.log(A["T_theory"])

# log T scale (no interaction): angle main effect, each angle mean over 9 runs (3 lengths x 3 blocks)
print("log T scale: angle levels (pooled over the lengths)")
tk_log = tukey_from_model(fact_anova_logT, A, "theta_deg", "logT")
th_log = A.groupby("theta_deg")["logT_theory"].mean()
tk_log["theory"] = [th_log[int(p.split(" - ")[0])] - th_log[int(p.split(" - ")[1])] for p in tk_log["pair"]]
tk_log["diff_pct"] = 100 * (np.exp(tk_log["diff"]) - 1) # effect in % of T
tk_log["theory_pct"] = 100 * (np.exp(tk_log["theory"]) - 1)
display(tk_log.round(5))

# T scale (interaction): compare the angles at each length, MS_E and df_E of the full T model, n = 3 per mean
rows = []
for L, g in A.groupby("L_m"):
    tk = tukey_from_model(fact_anova_T, g, "theta_deg", "T")
    th = g.groupby("theta_deg")["T_theory"].mean()
    tk["theory"] = [th[int(p.split(" - ")[0])] - th[int(p.split(" - ")[1])] for p in tk["pair"]]
    tk.insert(0, "L_m", L)
    rows.append(tk)
print("\nT scale: angle levels at each length")
display(pd.concat(rows, ignore_index=True).round(4))
```

Recap: Tukey's HSD controls the family-wise error rate for all three pairwise comparisons: $H_0: \mu_i = \mu_j$, the difference is significant if $|\bar y_i - \bar y_j| > q_{0.05;\,3,\,16}\sqrt{MS_E/n}$, with $MS_E$ and $df_E = 16$ of the blocked factorial model.

```python
# --- Task 5c: relative efficiency of the RCBD with respect to a CRD ---
def relative_efficiency(model, block, a, b):
    """Relative efficiency of an RCBD with a treatments
    and b blocks with respect to a CRD."""
    tab = sm.stats.anova_lm(model, typ=2)
    ms_block = tab.loc[block, "sum_sq"] / tab.loc[block, "df"]
    mse = model.mse_resid
    sigma2_crd = ((b - 1) * ms_block + b * (a - 1) * mse) / (a * b - 1)
    df_b, df_r = (a - 1) * (b - 1), a * (b - 1)
    correction = (df_b + 1) * (df_r + 3) / ((df_b + 3) * (df_r + 1))
    return sigma2_crd / mse, correction * sigma2_crd / mse

a_trt, b_blk = 9, A["block"].nunique() # 9 treatments (L x theta), 3 blocks (operators)
configurations = [("T", fact_anova_T), ("log T", fact_anova_logT)]

for name, fit in configurations:
    ratio, R = relative_efficiency(fit, "C(block)", a=a_trt, b=b_blk)
    print(f"{name:6s}: variance ratio = {ratio:.3f}, relative efficiency R = {R:.3f}")

# what the CRD analysis would give: the same model without blocks
crd_log = smf.ols("logT ~ C(L_m) * C(theta_deg)", data=A).fit()
print(
        f"\nMS_E with blocks = {fact_anova_logT.mse_resid:.3g} (df {fact_anova_logT.df_resid:.0f}), "
        f"without blocks = {crd_log.mse_resid:.3g} (df {crd_log.df_resid:.0f})\n"
        f"operator (block) means of log T: {A.groupby("operator")["logT"].mean().round(5).to_dict()}"
    )
```

means differ by only about 0.6 % (block $F = 0.89$, $p = 0.43$ on the $\log T$ scale), and without blocks $MS_E$ is even a bit smaller ($8.56e-5$ with 18 df vs $8.67e-5$ with 16 df), so on the $\log T$ scale we paid 2 error degrees of freedom for nothing.

*Limitation:*

Note that in Part A the three runs of one treatment were done one after another (timed by A, B, C), so the randomization within the blocks was restricted; the block test and the relative efficiency are therefore only a rough guide.

**Two squares, log scale.** The length dominates ($F = 7396$, $p < 10^{-7}$): 0.65 m and 1.0 m are $+0.300$ and

**One square only.** 9 runs and $1 + 4 \times 2 = 9$ parameters: the model is saturated, fits exactly ($R^2 = 1$) and

```python
# the same two-square model for T instead of log T, and the F-tests on both scales side by side
gl_T = smf.ols("T ~ " + F, data=B).fit()
aov_T = sm.stats.anova_lm(gl_T, typ=2)
comp = pd.concat({"log T": aov_log[["F", "PR(>F)"]], "T": aov_T[["F", "PR(>F)"]]}, axis=1).round(4)
print("F-tests on the two scales (two squares, residual df = 6):")
display(comp)

# aliasing in a 3x3 Graeco-Latin square: the angle x length interaction contrasts lie in the space of the row + mass contrasts
q = B[B["square"] == 1]

def dummies(d, cols):
    return pd.get_dummies(d[cols].astype(str)).astype(float).values

def resid_after(X, Y):                       # Y minus its projection on the column space of X
    return Y - X @ np.linalg.lstsq(X, Y, rcond=None)[0]

ones = np.ones((len(q), 1))
cells = dummies(q.assign(cell=q["theta_deg"].astype(str) + "_" + q["L_m"].astype(str)), ["cell"])  # 9 angle x length cells
inter = resid_after(np.hstack([ones, dummies(q, ["theta_deg", "L_m"])]), cells)   # interaction part of the cell means (4 df)
rowmass = np.hstack([ones, dummies(q, ["row", "n_rings"])])                        # row (operator) + mass main effects
print(f"rank of the angle x length interaction space in one square: {np.linalg.matrix_rank(inter)}; "
      f"largest residual after projecting it on the row + mass main effects: {np.abs(resid_after(rowmass, inter)).max():.1e}")

# consequence: push the exact theoretical periods (no noise, no mass effect, no operator effect) through the model
Bt = B.assign(T_th=theory_period(B["L_m"], B["theta_deg"]))
Bt["logT_th"] = np.log(Bt["T_th"])
ss = pd.concat({"theory T": sm.stats.anova_lm(smf.ols("T_th ~ " + F, data=Bt).fit(), typ=2)["sum_sq"],
                "theory log T": sm.stats.anova_lm(smf.ols("logT_th ~ " + F, data=Bt).fit(), typ=2)["sum_sq"]}, axis=1)
print("\nSums of squares when the response is the theoretical period itself (no noise):")
display(ss.apply(lambda c: c.map(lambda v: f"{v:.1e}")))

th = pd.DataFrame([(L, t) for L in sorted(B["L_m"].unique()) for t in sorted(B["theta_deg"].unique())],
                  columns=["L_m", "theta_deg"])
th["T"] = theory_period(th["L_m"], th["theta_deg"])
pv = th.pivot(index="L_m", columns="theta_deg", values="T")
print("theoretical angle effect 45 deg - 5 deg:  on T [s]:", (pv[45] - pv[5]).round(4).to_dict(),
      "  on log T:", (np.log(pv[45]) - np.log(pv[5])).round(5).to_dict())
```

**$T$ versus $\log T$.** Almost every $F$ is larger on the $\log T$ scale (mass $3.93$ vs $3.65$, angle $23.34$ vs $21.79$, length
$7396$ vs $6650$, operators $2.84$ vs $2.25$), with the same decisions, but those tests are the wrong ones.

**Aliasing.** The 9 cells of a $3 \times 3$ square carry 8 df and the four factors use all of them: the design is
saturated. An angle $\times$ length interaction (4 df) is orthogonal to both main effects, so it must live in the
remaining 4 dimensions, which are exactly the operator and mass effects (the check gives rank 4 and residual
$10^{-15}$ after projection on them). Any interaction is therefore **completely confounded** with the operator and mass
estimates.

On the $T$ scale angle and length multiply (the 45° to 5° difference is 0.047 s at 0.35 m, 0.079 s at 1 m). Feeding the
exact theoretical periods through the model gives spurious mass and operator effects ($SS = 1.7 \times 10^{-6}$ and
$2.6 \times 10^{-4}$, about 0.3 %) plus $2.6 \times 10^{-4}$ in the residual, all of which vanish to machine precision
on the $\log T$ scale, where the model is exactly additive
($\log T = \text{const} + \tfrac12 \log L + x(\theta_0)$, no mass term).

  $\hat\sigma = 0.7\,\%$.
* One square: 9 parameters for 9 runs, 0 residual df, no test; the second square supplies the error term.
* Use $\log T$: on the $\log T$ scale the model is additive.

**2**
* Choose an equivalence margin $\Delta$ (in seconds or in % of $T$) **and justify it** (e.g. the precision needed to estimate $g$
  to 1 %). Test $H_0: |\text{effect}| \ge \Delta$ against $H_1: |\text{effect}| < \Delta$ by two one-sided $t$-tests (TOST) with
  $\alpha = 0.05$; equivalently, check whether the 90 % confidence interval lies inside $(-\Delta, \Delta)$.

```python
# Equivalence margin

delta_1pct = 0.01
delta_2pct = 0.02

print("=== Equivalence margin with alpha = 0.05 ===\n")

t_val_b_90 = stats.t.ppf(0.95, df_resid_b)  # 90% CI critical value for factor model
t_val_num_90 = stats.t.ppf(0.95, df_resid_num)  # 90% CI critical value for numeric model

# 1. Factor model
factor_ci90_low = mass3_diff - t_val_b_90 * mass3_se
factor_ci90_high = mass3_diff + t_val_b_90 * mass3_se

print("1. Factor Model:")
print(f"   Estimated Effect:  {mass3_diff:.5f} ({100 * (np.exp(mass3_diff)-1):+.2f}%)")
print(f"   90% CI on logT:    ({factor_ci90_low:.5f}, {factor_ci90_high:.5f})")
print(f"   90% CI in % of T:  ({100 * (np.exp(factor_ci90_low)-1):+.2f}%, {100 * (np.exp(factor_ci90_high)-1):+.2f}%)")

for d, label in [(delta_1pct, "1.0%"), (delta_2pct, "2.0%")]:
    equiv = (factor_ci90_low > -d) and (factor_ci90_high < d)
    print(f"   Is effect equivalent within {label} margin? {equiv}")

# 2. Numeric model
num_ci90_low = effect_range - t_val_num_90 * effect_se_range
num_ci90_high = effect_range + t_val_num_90 * effect_se_range

print("\n2. Numeric model:")
print(f"   Estimated Effect:  {effect_range:.5f} ({100 * (np.exp(effect_range)-1):+.2f}%)")
print(f"   90% CI on logT:    ({num_ci90_low:.5f}, {num_ci90_high:.5f})")
print(f"   90% CI in % of T:  ({100 * (np.exp(num_ci90_low)-1):+.2f}%, {100 * (np.exp(num_ci90_high)-1):+.2f}%)")

for d, label in [(delta_1pct, "1.0%"), (delta_2pct, "2.0%")]:
    equiv = (num_ci90_low > -d) and (num_ci90_high < d)
    print(f"   Is effect equivalent within {label} margin? {equiv}")
```

#### Choice of the equivalence margin ($\Delta$) and its justification:
* **Physical Basis:** The acceleration of gravity is $g = (2\pi)^2 L / T^2$, which implies $\log g = \log(2\pi)^2 + \log L - 2\log T$. Therefore, a small relative error in the period ($dT/T$) translates to twice that relative error in our estimate of $g$ ($dg/g \approx -2 \cdot dT/T$).
* **$\Delta = 1.0\%$:** if we require $g$ to be estimated to within **2.0%** precision, the period deviation due to mass must be strictly under **1.0%** ($|\text{effect}| < 0.01$ on the log scale).
* **$\Delta = 2.0\%$:** if we are more lenient and require $g$ to be estimated to within **4.0%** precision, the period deviation due to mass must be under **2.0%** ($|\text{effect}| < 0.02$ on the log scale).

#### Test hypotheses and results
To investigate whether the mass effect is statistically negligible, we perform two one-sided tests at significance level $\alpha = 0.05$. This is equivalent to checking if the **90% confidence interval** of the estimated mass effect lies entirely inside the interval $(-\Delta, \Delta)$.

* **For $\Delta = 1.0\%$:**
  both 90% CIs extend beyond $+1.0\%$, so we **cannot** reject the null hypothesis of non-equivalence. We cannot claim statistical equivalence under this strict margin.
* **For $\Delta = 2.0\%$:**
  In this case, both 90% CIs lie entirely within $[-2.0\%, +2.0\%]$, therefore, we **reject** the null hypothesis of non-equivalence ($p < 0.05$ for both one-sided tests). We can claim that the mass effect is statistically equivalent to zero.

**3**
* Explain the difference between "the mass effect is not significant" and "the mass effect is negligible".

-  "The mass effect is not significant" means failing to reject the null hypothesis $H_0: \text{effect} = 0$, that is we lack sufficient statistical evidence or power to distinguish the estimated effect of the mass from zero. It does **not** prove that the effect is actually zero. The $95\%$ confidence intervals included $0\%$, but they also contained values as high as $+2.1\%$, which could still represent a physically meaningful effect.

-  "The mass effect is negligible" means rejecting the null hypothesis $H_0: |\text{effect}| \ge \Delta$, that is the statistical precision is high enough to confidently state that the entire $90\%$ confidence interval of the effect lies strictly inside the boundaries $(-\Delta, \Delta)$. It provides statistical evidence that even if a physical mass effect exists, it is guaranteed to be too small to matter for this application.

```python
# --- Task 9: model diagnostics (log T scale, chosen in Task 5) ---
def diagnostics(fit, data, title, factors):
    """Residuals vs fitted, normal Q-Q plot, residuals vs each factor, vs operator and vs run order."""
    d = data.loc[fit.resid.index].copy()
    d["resid"], d["fitted"] = fit.resid, fit.fittedvalues
    panels = ["fitted", "qq"] + factors + ["operator", "run"]
    fig, axes = plt.subplots(2, 4, figsize=(17, 7))
    axes = axes.ravel()
    for ax, p in zip(axes, panels):
        if p == "qq":
            sm.qqplot(d["resid"], line="s", ax=ax)
            ax.set_title("Normal Q-Q plot")
            continue
        if p in factors + ["operator"]:              # categorical: sorted levels at positions 0, 1, 2
            levels = sorted(d[p].unique())
            x = d[p].map({v: i for i, v in enumerate(levels)})
            ax.set_xticks(range(len(levels)), [str(v) for v in levels])
        else:
            x = d[p]
        ax.scatter(x, d["resid"], alpha=0.7)
        ax.axhline(0, color="grey", ls="--")
        ax.set(xlabel=p, ylabel="residual (log T)", title=f"Residuals vs {p}")
    for ax in axes[len(panels):]:
        ax.set_visible(False)
    fig.suptitle(title, fontsize=13)
    plt.tight_layout()
    plt.show()
```

### 10. Comparison of the two designs

**1**
* For the angle effect (45° - 5°) and the length effect on the $\log T$ scale, compare the standard errors from Part A and

```python
cov_a = fact_anova_logT.cov_params()
cov_b = gl_log.cov_params()

n_a = len(A)
n_b = len(B)

print(f"Number of runs: Part A = {n_a}, Part B = {n_b}\n")

effects = {
    "Angle (45° vs 5°)": ("C(theta_deg)[T.45]", "C(theta_deg)[T.45]"),
    "Length (0.65m vs 0.35m)": ("C(L_m)[T.0.65]", "C(L_m)[T.0.65]"),
    "Length (1.00m vs 0.35m)": ("C(L_m)[T.1.0]", "C(L_m)[T.1.0]")
}

comparison_rows = []
for name, (param_a, param_b) in effects.items():
    var_a = cov_a.loc[param_a, param_a]
    se_a = np.sqrt(var_a)
    per_run_var_a = var_a * n_a

    var_b = cov_b.loc[param_b, param_b]
    se_b = np.sqrt(var_b)
    per_run_var_b = var_b * n_b

    comparison_rows.append({
        "Effect": name,
        "Part A SE": se_a,
        "Part B SE": se_b,
        "Part A Per-Run Var": per_run_var_a,
        "Part B Per-Run Var": per_run_var_b,
        "Efficiency Ratio (B/A)": per_run_var_b / per_run_var_a if per_run_var_a > 0 else np.nan
    })

comp_df = pd.DataFrame(comparison_rows)
display(comp_df.round(6))
```

Part A can estimate the interactions $L \theta_0$, indeed every treatment combination of Length and Angle is tested across all replicates; clearly it cannot estimate the mass effect since the mass is held constant (at 2 rings) throughout all runs of Part A.

On the other hand, Part B cannot estimate any interaction since there are not enough degrees of freedom, but it can estimate multiple main effects simultaneously, indeed, in only 18 runs (across two squares), it can estimate the main effects of four factors: Length ($L$), Angle ($\theta_0$), Mass (number of rings), and Operator (rows).

Part A assumptions:
Block-Treatment Additivity: the effect of the operators (blocks) is strictly additive and does not interact with length or angle; our data support this assumption as the relative efficiency calculation ($R \approx 0.98$ to $1.05$) showed that the operator variances were extremely small, confirming that blocking was neutral and did not suffer from massive interaction distortions.

Part B Assumptions:
Strict Multi-way Additivity: zero interactions between any of the four factors (Length, Angle, Mass, Operator); in this case this assumption is not supported by the data, indeed physically the raw period $T$ is non-linear ($T_0 \approx 2\pi\sqrt{L/g}$). On a raw seconds scale, the angle effect increases with length, violating the assumption of zero interactions. We tried to overcome this by introducing the logarithmic scale, which transforms the multiplicative physical relationships into additive terms: $$\log T = \log(2\pi) - \frac{1}{2}\log g + \frac{1}{2}\log L + x(\theta_0)$$
